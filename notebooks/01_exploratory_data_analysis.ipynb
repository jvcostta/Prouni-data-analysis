# 📊 Análise Exploratória dos Dados do ProUni (2019-2020)

Este notebook contém a análise exploratória inicial dos dados do Programa Universidade para Todos (ProUni) para os anos de 2019 e 2020.

## 🎯 Objetivos
1. **Carregar e explorar** a estrutura dos dados
2. **Identificar** tipos de dados e possíveis inconsistências
3. **Analisar** dados faltantes
4. **Gerar** estatísticas descritivas básicas
5. **Verificar** se os datasets de diferentes anos têm a mesma estrutura
6. **Criar** visualizações iniciais para entender os padrões

## 📚 Bibliotecas

```python
# Importação das bibliotecas necessárias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings

# Configurações
warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8')
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

print("✅ Bibliotecas importadas com sucesso!")
```

## 📁 1. Carregamento dos Dados

```python
# Caminhos para os arquivos
arquivo_2019 = '../data/raw/pda-prouni-2019.csv'
arquivo_2020 = '../data/raw/ProuniRelatorioDadosAbertos2020.csv'

# Tentativa de carregamento com diferentes encodings
def carregar_csv_seguro(caminho, nome_ano):
    encodings = ['utf-8', 'latin-1', 'iso-8859-1', 'cp1252']
    
    for encoding in encodings:
        try:
            df = pd.read_csv(caminho, encoding=encoding, sep=';')
            print(f"✅ {nome_ano}: Carregado com encoding {encoding}")
            print(f"   Shape: {df.shape}")
            return df
        except UnicodeDecodeError:
            continue
        except Exception as e:
            print(f"❌ Erro ao carregar {nome_ano} com {encoding}: {e}")
            continue
    
    # Tentativa com vírgula como separador
    for encoding in encodings:
        try:
            df = pd.read_csv(caminho, encoding=encoding, sep=',')
            print(f"✅ {nome_ano}: Carregado com encoding {encoding} e separador vírgula")
            print(f"   Shape: {df.shape}")
            return df
        except:
            continue
    
    print(f"❌ Não foi possível carregar {nome_ano}")
    return None

# Carregamento dos dados
print("🔄 Carregando dados...")
df_2019 = carregar_csv_seguro(arquivo_2019, "ProUni 2019")
df_2020 = carregar_csv_seguro(arquivo_2020, "ProUni 2020")
```

## 🔍 2. Primeira Exploração - Estrutura dos Dados

```python
# Função para explorar estrutura básica
def explorar_estrutura(df, nome):
    print(f"\n{'='*50}")
    print(f"📊 ANÁLISE: {nome}")
    print(f"{'='*50}")
    
    if df is not None:
        print(f"\n📏 Dimensões: {df.shape[0]:,} linhas x {df.shape[1]} colunas")
        
        print(f"\n📋 Colunas disponíveis:")
        for i, col in enumerate(df.columns, 1):
            print(f"  {i:2d}. {col}")
        
        print(f"\n🔢 Tipos de dados:")
        print(df.dtypes.value_counts())
        
        print(f"\n👀 Primeiras 3 linhas:")
        display(df.head(3))
        
        return True
    else:
        print(f"❌ Dados não disponíveis para {nome}")
        return False

# Explorar ambos os datasets
sucesso_2019 = explorar_estrutura(df_2019, "ProUni 2019")
sucesso_2020 = explorar_estrutura(df_2020, "ProUni 2020")
```

## 🔎 3. Comparação da Estrutura entre Anos

```python
# Comparar estruturas se ambos os datasets foram carregados
if sucesso_2019 and sucesso_2020:
    print("🔄 Comparando estruturas dos datasets...\n")
    
    # Comparar colunas
    cols_2019 = set(df_2019.columns)
    cols_2020 = set(df_2020.columns)
    
    cols_comuns = cols_2019.intersection(cols_2020)
    cols_apenas_2019 = cols_2019 - cols_2020
    cols_apenas_2020 = cols_2020 - cols_2019
    
    print(f"📊 Resumo da Comparação:")
    print(f"  • Colunas em comum: {len(cols_comuns)}")
    print(f"  • Apenas em 2019: {len(cols_apenas_2019)}")
    print(f"  • Apenas em 2020: {len(cols_apenas_2020)}")
    
    if cols_apenas_2019:
        print(f"\n🔹 Colunas apenas em 2019:")
        for col in sorted(cols_apenas_2019):
            print(f"  • {col}")
    
    if cols_apenas_2020:
        print(f"\n🔹 Colunas apenas em 2020:")
        for col in sorted(cols_apenas_2020):
            print(f"  • {col}")
    
    print(f"\n✅ Colunas comuns: {len(cols_comuns)}")
    for col in sorted(cols_comuns):
        print(f"  • {col}")
else:
    print("❌ Não é possível comparar - algum dataset não foi carregado")
```

## 📈 4. Estatísticas Descritivas

```python
# Função para gerar estatísticas descritivas
def estatisticas_descritivas(df, nome):
    if df is not None:
        print(f"\n{'='*30} {nome} {'='*30}")
        
        # Informações gerais
        print(f"\n📊 Informações Gerais:")
        print(df.info())
        
        # Estatísticas para colunas numéricas
        print(f"\n🔢 Estatísticas Numéricas:")
        display(df.describe())
        
        # Estatísticas para colunas categóricas
        print(f"\n📝 Estatísticas Categóricas (primeiras colunas text):")
        colunas_texto = df.select_dtypes(include=['object']).columns[:5]
        if len(colunas_texto) > 0:
            display(df[colunas_texto].describe())

# Gerar estatísticas para ambos os datasets
if sucesso_2019:
    estatisticas_descritivas(df_2019, "ProUni 2019")

if sucesso_2020:
    estatisticas_descritivas(df_2020, "ProUni 2020")
```

## 🕳️ 5. Análise de Dados Faltantes

```python
# Função para analisar dados faltantes
def analisar_dados_faltantes(df, nome):
    if df is not None:
        print(f"\n{'='*20} DADOS FALTANTES - {nome} {'='*20}")
        
        # Contagem de valores nulos
        nulos = df.isnull().sum()
        nulos_percent = (nulos / len(df)) * 100
        
        # Criar DataFrame com informações de dados faltantes
        dados_faltantes = pd.DataFrame({
            'Coluna': nulos.index,
            'Valores_Nulos': nulos.values,
            'Percentual': nulos_percent.values
        })
        
        # Filtrar apenas colunas com dados faltantes
        dados_faltantes = dados_faltantes[dados_faltantes['Valores_Nulos'] > 0]
        dados_faltantes = dados_faltantes.sort_values('Percentual', ascending=False)
        
        if len(dados_faltantes) > 0:
            print(f"\n❌ Colunas com dados faltantes:")
            display(dados_faltantes)
            
            # Visualização dos dados faltantes
            if len(dados_faltantes) <= 20:  # Só plotar se não houver muitas colunas
                plt.figure(figsize=(12, 6))
                plt.bar(dados_faltantes['Coluna'], dados_faltantes['Percentual'])
                plt.title(f'Percentual de Dados Faltantes - {nome}')
                plt.xlabel('Colunas')
                plt.ylabel('Percentual (%)')
                plt.xticks(rotation=45, ha='right')
                plt.tight_layout()
                plt.show()
        else:
            print(f"✅ Nenhum dado faltante encontrado!")
        
        return dados_faltantes

# Analisar dados faltantes
if sucesso_2019:
    faltantes_2019 = analisar_dados_faltantes(df_2019, "ProUni 2019")

if sucesso_2020:
    faltantes_2020 = analisar_dados_faltantes(df_2020, "ProUni 2020")
```

## 📊 6. Visualizações Iniciais

Vamos criar algumas visualizações básicas para entender melhor os dados.

```python
# Função para criar visualizações básicas
def visualizacoes_basicas(df, nome, ano):
    if df is not None:
        print(f"\n📈 Criando visualizações para {nome}...")
        
        # Tentar identificar colunas relevantes
        colunas = df.columns.tolist()
        
        # Buscar por colunas que possam representar dados importantes
        col_uf = None
        col_modalidade = None
        col_bolsa = None
        col_curso = None
        
        for col in colunas:
            col_lower = col.lower()
            if 'uf' in col_lower or 'estado' in col_lower:
                col_uf = col
            elif 'modalidade' in col_lower:
                col_modalidade = col
            elif 'bolsa' in col_lower or 'tipo' in col_lower:
                col_bolsa = col
            elif 'curso' in col_lower:
                col_curso = col
        
        # Visualização 1: Distribuição por UF (se encontrada)
        if col_uf:
            plt.figure(figsize=(15, 8))
            uf_counts = df[col_uf].value_counts().head(15)
            plt.bar(uf_counts.index, uf_counts.values)
            plt.title(f'Top 15 Estados - {nome}')
            plt.xlabel('UF')
            plt.ylabel('Quantidade de Registros')
            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()
        
        # Visualização 2: Distribuição por Modalidade (se encontrada)
        if col_modalidade:
            plt.figure(figsize=(10, 6))
            modalidade_counts = df[col_modalidade].value_counts()
            plt.pie(modalidade_counts.values, labels=modalidade_counts.index, autopct='%1.1f%%')
            plt.title(f'Distribuição por Modalidade - {nome}')
            plt.show()
        
        # Visualização 3: Top cursos (se encontrada)
        if col_curso:
            plt.figure(figsize=(15, 8))
            curso_counts = df[col_curso].value_counts().head(20)
            plt.barh(range(len(curso_counts)), curso_counts.values)
            plt.yticks(range(len(curso_counts)), curso_counts.index)
            plt.title(f'Top 20 Cursos - {nome}')
            plt.xlabel('Quantidade de Registros')
            plt.tight_layout()
            plt.show()
        
        print(f"✅ Visualizações criadas para {nome}")
        
        return {
            'col_uf': col_uf,
            'col_modalidade': col_modalidade,
            'col_bolsa': col_bolsa,
            'col_curso': col_curso
        }

# Criar visualizações
if sucesso_2019:
    colunas_2019 = visualizacoes_basicas(df_2019, "ProUni 2019", 2019)

if sucesso_2020:
    colunas_2020 = visualizacoes_basicas(df_2020, "ProUni 2020", 2020)
```
